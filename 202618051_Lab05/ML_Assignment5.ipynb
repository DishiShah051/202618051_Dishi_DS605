```python
def train_logistic_regression(X, y, learning_rate=0.01, iterations=5000):
    weights = np.zeros(X.shape[1])
    start_time = time.perf_counter()

    for i in range(iterations):
        probabilities = sigmoid(X @ weights)
        gradient = (X.T @ (probabilities - y)) / len(y)
        weights -= learning_rate * gradient
    training_time = time.perf_counter() - start_time
    return weights, training_time
```

```python
def predict_logistic_regression(X, weights, threshold=0.5):
    probabilities = sigmoid(X @ weights)
    predictions = (probabilities >= threshold).astype(int)
    return predictions
```

```python
def classification_metrics_manual(y_true, y_pred):
    TP = np.sum((y_true == 1) & (y_pred == 1))
    TN = np.sum((y_true == 0) & (y_pred == 0))
    FP = np.sum((y_true == 0) & (y_pred == 1))
    FN = np.sum((y_true == 1) & (y_pred == 0))

    accuracy = (TP + TN) / (TP + TN + FP + FN)
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall = TP / (TP + FN) if (TP + FN) > 0 else 0
    f1 = (2 * precision * recall / (precision + recall) if (precision + recall) > 0 else 0)
    return accuracy, precision, recall, f1
```

```python
def train_logistic_optimized(X, y, learning_rate=0.01, max_iterations=2000, tolerance=1e-7
):
    weights = np.zeros(X.shape[1])
    n = len(y)
    start_time = time.perf_counter()
    previous_loss = float("inf")

    for i in range(max_iterations):
        # Predictions
        probabilities = sigmoid(X @ weights)
        # Gradient
        gradient = (X.T @ (probabilities - y)) / n
        # Update weights
        weights -= learning_rate * gradient
        # Calculate loss
        probabilities = np.clip(sigmoid(X @ weights),1e-15,1 - 1e-15)
        loss = -np.mean(y * np.log(probabilities)+ (1 - y) * np.log(1 - probabilities))

        # Check convergence
        if abs(previous_loss - loss) < tolerance:
            break
        previous_loss = loss
    training_time = time.perf_counter() - start_time
    return weights, training_time, i + 1, loss
```

```python
optimized_weights, optimized_time, iterations_used, final_loss = (train_logistic_optimized(X_cls_train_manual, y_cls_manual_train, learning_rate=0.01, max_iterations=2000,
tolerance=1e-7))
optimized_predictions = predict_logistic_regression(X_cls_test_manual, optimized_weights)
optimized_accuracy, optimized_precision, optimized_recall, optimized_f1 = (classification_metrics_manual(y_cls_manual_test, optimized_predictions))
print("Iterations used:", iterations_used)
print("Final loss:", final_loss)
print("Training time:", optimized_time)
print("Accuracy:", optimized_accuracy)
print("Precision:", optimized_precision)
print("Recall:", optimized_recall)
print("F1:", optimized_f1)
```

```python
final_learning_rate = 0.01
final_iterations = 2000
start_time = time.perf_counter()
final_weights = np.zeros(X_cls_train_manual.shape[1])

for i in range(final_iterations):
    probabilities = sigmoid(X_cls_train_manual @ final_weights)
    gradient = (X_cls_train_manual.T @ (probabilities - y_cls_manual_train)) / len(y_cls_manual_train)
    final_weights -= final_learning_rate * gradient
final_optimized_training_time = (time.perf_counter() - start_time)
```
