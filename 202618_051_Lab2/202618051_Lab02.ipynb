## **Task 2-Vectorized Arithmetic and Linear Algebra**

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
print("Addition:",A+B)
print("Element-wise Multiplication:",A*B)
print("Matrix Multiplication:",A@B)
```

```python
A=np.array([[1,2,3],[4,5,6],[1,3,9]])
print("Transpose:",A.T)
print("Determinant:",np.linalg.det(A))
```

```python
A=np.array([[4,7],[2,6]])
det=np.linalg.det(A)
if det != 0:
    A_inv=np.linalg.inv(A)
    print("Inverse:",A_inv)
    print("Verified:",np.allclose(A @ A_inv,np.eye(A.shape[0])))
else:
    print("Matrix is not invertible.")
```

## **Task 3-Normal Distribution and Histogram**

```python
mean=50
std=10
data=np.random.normal(mean,std,1000)
print("Mean:",mean)
print("Standard Deviation:",std)
print("Generated values:",data)
```

```python
sample_mean=np.mean(data)
sample_std=np.std(data,ddof=1)
print("Sample Mean:",sample_mean)
print("Sample Standard Deviation:",sample_std)
print("Mean Difference:",sample_mean-50)
print("Standard Deviation Difference:",sample_std-10)
```

# **Part B-Data Wrangling with Pandas: Titanic Dataset**

## **Task 4-Load and Inspect Data**

```python
import pandas as pd
df=pd.read_csv("train.csv")
```

```python
print("Head:\n",df.head())
print("\nTail:\n",df.tail())
print("\nShape:",df.shape)
print("\nColumns:",df.columns)
print("\nInfo:",df.info())
print("\nDescription:",df.describe())
```

## **Task 5-Filtering and Querying**

```python
male_over_50=df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("Male passengers older than 50:", len(male_over_50))

female_first=df[(df["Sex"] == "female") & (df["Pclass"] == 1)]
survival_percentage=female_first["Survived"].mean()*100
print("\nFemale first-class passengers:", len(female_first))
print("Percentage survived:", survival_percentage)

fare_median=df["Fare"].median()
age_fare_survived=df[(df["Age"].between(20, 40)) & (df["Fare"] > fare_median) & (df["Survived"] == 1)]
print("\nAge 20-40, fare above median, survived:", len(age_fare_survived))

alone_under_30=df[(df["SibSp"] == 0) & (df["Parch"] == 0) & (df["Age"] < 30) & (df["Survived"] == 0)]
print("\nTravelling alone, age < 30, did not survive:", len(alone_under_30))

southampton_median=df.loc[df["Embarked"] == "S", "Fare"].median()
southampton_filter=df[(df["Embarked"] == "S") & (df["Pclass"].isin([2, 3])) & (df["Fare"] > southampton_median)]
print("\nEmbarked S, Pclass 2/3, fare above Southampton median:",len(southampton_filter))
```

## **Task 6-groupby() and Aggregation**

```python
print("Survival Rate by Sex:")
print(df.groupby("Sex")["Survived"].mean())

print("\nSurvival Rate by Pclass:")
print(df.groupby("Pclass")["Survived"].mean())

print("\nAverage Age and Fare by Pclass:")
print(df.groupby("Pclass")[["Age", "Fare"]].mean())

print("\nPassenger Count and Survival Rate by Sex-Pclass:")
print(df.groupby(["Sex", "Pclass"]).agg(Passenger_Count=("PassengerId", "count"),Survival_Rate=("Survived", "mean")))

print("\nPassenger Count, Average Fare, and Survival Rate by Embarked:")
print(df.groupby("Embarked").agg(Passenger_Count=("PassengerId", "count"),Average_Fare=("Fare", "mean"),Survival_Rate=("Survived", "mean")))
```

## **Task 7-Missing Values and Fare Outliers**

```python
missing_count=df.isnull().sum()
missing_percentage=(missing_count / len(df))*100
missing_summary=pd.DataFrame({"Missing Count": missing_count,"Missing Percentage": missing_percentage})
print("Missing Values:")
print(missing_summary)
missing_count.plot(kind="bar")
plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)
plt.show()
plt.savefig("Missing_Values.png")
```

```python
print("Missing Age before:", df["Age"].isnull().sum())
mean_age=df["Age"].mean()
df["Age"]=df["Age"].fillna(mean_age)
print("Missing Age after:", df["Age"].isnull().sum())
```

```python
df_mean=pd.read_csv("train.csv")
df_median=pd.read_csv("train.csv")
df_mode=pd.read_csv("train.csv")
df_random=pd.read_csv("train.csv")
df_mean["Age"]=df_mean["Age"].fillna(df_mean["Age"].mean())
df_median["Age"]=df_median["Age"].fillna(df_median["Age"].median())
mode_age=df_mode["Age"].mode()[0]
df_mode["Age"]=df_mode["Age"].fillna(mode_age)
np.random.seed(42)
random_values=np.random.choice(df_random["Age"].dropna(), size=df_random["Age"].isnull().sum())
df_random.loc[df_random["Age"].isnull(), "Age"]=random_values
print("\nMissing Age after Mean Imputation:", df_mean["Age"].isnull().sum())
print("Missing Age after Median Imputation:", df_median["Age"].isnull().sum())
print("Missing Age after Mode Imputation:", df_mode["Age"].isnull().sum())
print("Missing Age after Random Imputation:", df_random["Age"].isnull().sum())
```

```python
Q1=df["Fare"].quantile(0.25)
Q3=df["Fare"].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
outliers=df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]
print("Fare Outlier Analysis:")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

## **Task 8-Features and Pivot Table**

```python
df["FamilySize"]=df["SibSp"]+df["Parch"]+1
df["IsAlone"]=(df["FamilySize"] == 1).astype(int)
print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head())
```

```python
pivot=pd.pivot_table(df, index="Sex", columns="Pclass", values="Survived", aggfunc="mean")
print("Pivot Table:")
print(pivot)
highest=pivot.stack().idxmax()
lowest=pivot.stack().idxmin()
print("Highest survival group:", highest)
print("Highest survival rate:", pivot.loc[highest[0], highest[1]])
print("Lowest survival group:", lowest)
print("Lowest survival rate:", pivot.loc[lowest[0], lowest[1]])
```

## **Task 9-Visualizations and Observations**

```python
import seaborn as sns
numerical_cols=["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
corr=df[numerical_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
corr_pairs=corr.where(~corr.eq(1))
corr_pairs=corr_pairs.stack()
print("Strongest positive relationship:")
print(corr_pairs.idxmax(), corr_pairs.max())
print("Strongest negative relationship:")
print(corr_pairs.idxmin(), corr_pairs.min())
plt.savefig("Correlation_Heatmap.png")
```

```python
survival_rate=df.groupby("Sex")["Survived"].mean()*100
print("Male survival rate:", survival_rate["male"], "%")
print("Female survival rate:", survival_rate["female"], "%")
survival_rate.plot(kind="bar")
plt.title("Survival Rate by Sex")
plt.xlabel("Sex")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=0)
plt.show()
plt.savefig("Survival_Rate.png")
```

```python
for survival in [0, 1]:
    data=df[df["Survived"] == survival]
    plt.scatter(data["Age"], data["Fare"], label=f"Survived={survival}")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare by Survival")
plt.legend()
plt.show()
plt.savefig("Age_VS_Fare_Survival.png")
```

**Key Observations**
*   Female passengers had a higher survival rate than male passengers.
*   First-class passengers generally had a higher survival rate than second- and third-class passengers.
* Passengers with higher fares were more likely to belong to higher passenger classes.
* The Age column contained missing values, which were handled using imputation.
* The Fare column contained outliers, identified using the 1.5 × IQR rule.
* Passengers travelling alone showed different survival patterns compared with passengers travelling with family.
* The Age vs Fare plot shows that fare values were widely spread, while most passengers were concentrated in the lower-to-middle age range.
